````python
nombre = input("Nombre: ")
peso = float(input("Peso (kg): "))
altura = float(input("Altura (m): "))
imc = peso / (altura ** 2)

# Clasificación
if imc < 18.5:
    cat, msg = "Bajo peso", "Debes aumentar tu ingesta calórica."
elif imc < 25:
    cat, msg = "Normal", "¡Sigue así! Mantienes un buen equilibrio."
elif imc < 30:
    cat, msg = "Sobrepeso", "Se recomienda actividad física regular."
else:
    cat, msg = "Obesidad", "Es importante consultar a un médico."

print(f"\n{nombre}, tu IMC es {imc:.2f}. Categoría: {cat}. {msg}")
````

````python
n1 = float(input("Número 1: "))
n2 = float(input("Número 2: "))
op = input("Operación (+, -, *, /): ")

if op == "+": res = n1 + n2
elif op == "-": res = n1 - n2
elif op == "*": res = n1 * n2
elif op == "/": res = n1 / n2 if n2 != 0 else "Error: División por cero"
else: res = "Operación no válida"

print(f"Resultado: {res}")
````
